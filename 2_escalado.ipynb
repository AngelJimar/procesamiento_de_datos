# Escalado
Conjunto de datos: iris (incluido en sklearn)
Este conjunto de datos es compacto y perfecto para demostrar técnicas de escalado.

```python
# Importar librerías necesarias
from sklearn.datasets import load_iris
from sklearn.preprocessing import MinMaxScaler, StandardScaler
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Cargar el conjunto de datos
iris = load_iris(as_frame=True)
data = iris.data
```

```python
# Visualizar las estadísticas iniciales
print("Estadísticas iniciales:")
data.describe()
```

```python
# Aplicar MinMaxScaler
minmax_scaler = MinMaxScaler()
data_minmax = pd.DataFrame(minmax_scaler.fit_transform(data), columns=data.columns)
```

```python
# Aplicar StandardScaler
standard_scaler = StandardScaler()
data_standard = pd.DataFrame(standard_scaler.fit_transform(data), columns=data.columns)
```

### 2. Crea una gráfica de la distribución de la variable 'sepal length (cm)' antes y después del escalado.

```python
# Configuracion para gráficas paralelas
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Distribución antes del escalado (Original)
axes[0].hist(data['sepal length (cm)'], bins=15, color='blue', alpha=0.7, edgecolor='black')
axes[0].set_title('Original: Sepal Length')
axes[0].set_xlabel('Centímetros')
axes[0].set_ylabel('Frecuencia')

# Distribución después del escalado (MinMax)
axes[1].hist(data_minmax['sepal length (cm)'], bins=15, color='red', alpha=0.7, edgecolor='black')
axes[1].set_title('MinMax Scaled: Sepal Length')
axes[1].set_xlabel('Valor Escalado (0 a 1)')
axes[1].set_ylabel('Frecuencia')

plt.suptitle('Comparación de Distribución Antes y Después del Escalado', fontsize=14)
plt.tight_layout()
plt.show()
```

### 3. Crea una gráfica de dispersión entre 'sepal length (cm)' y 'sepal width (cm)' con los datos escalados por StandardScaler.

```python
# Grafica los datos escalados por StandardScaler
plt.figure(figsize=(8, 6))
plt.scatter(data_standard['sepal length (cm)'], 
            data_standard['sepal width (cm)'], 
            color='blue', alpha=0.8, edgecolors='black')

# Personaliza los detalles de la gráfica
plt.title('Gráfica de Dispersión con Datos Escalados (StandardScaler)', fontsize=12)
plt.xlabel('Sepal Length (Estandarizado: Media ~ 0, STD = 1)')
plt.ylabel('Sepal Width (Estandarizado: Media ~ 0, STD = 1)')
plt.grid(True, linestyle='--', alpha=0.5)

plt.show() 
```
